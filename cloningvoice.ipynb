```python
%%writefile /content/voice_agent.py
import os
os.environ['MPLBACKEND'] = 'Agg'

from TTS.api import TTS
import torch
import gradio as gr

# Load model once (stays in memory)
print("Loading voice model...")
device = "cuda" if torch.cuda.is_available() else "cpu"
tts = TTS('tts_models/multilingual/multi-dataset/xtts_v2').to(device)
print("Model ready!")

# Your voice file
YOUR_VOICE = "/content/drive/MyDrive/my_voice_dataset/arabic/001.wav"

def speak(text, language):
    """
    Generate speech in your voice for any text and language
    """
    # Map language names to XTTS language codes
    lang_map = {
        "Arabic": "ar",
        "English": "en",
        "French": "fr"
    }

    lang_code = lang_map[language]

    # Generate audio file
    output_path = f"/content/output_{lang_code}.wav"
    tts.tts_to_file(
        text=text,
        speaker_wav=YOUR_VOICE,
        language=lang_code,
        file_path=output_path
    )

    return output_path

# Create web interface
interface = gr.Interface(
    fn=speak,
    inputs=[
        gr.Textbox(label="Enter text to speak", placeholder="Type anything..."),
        gr.Dropdown(choices=["Arabic", "English", "French"], label="Select language")
    ],
    outputs=gr.Audio(label="Your voice speaking", type="filepath"),
    title="🎙️ Voice Cloning Agent",
    description="Type any text and select a language. The AI will speak it in MY voice!",
    examples=[
        ["مرحبا كيف حالك", "Arabic"],
        ["Hello how are you today", "English"],
        ["Bonjour tout le monde", "French"],
        ["أنا أحب التعلم الآلي", "Arabic"],
        ["Artificial intelligence is amazing", "English"],
        ["L'intelligence artificielle est fascinante", "French"]
    ]
)

# Launch the app
if __name__ == "__main__":
    interface.launch(share=True)
```
